```python
Setting = pypolo2.utilities.Config()
strategy_name = ["EffectOrientedSelectiveSpray",
                 "EffectOrientedNonDecoupledSpray",
               #   "EffectOrientedMCTSNonMyopicSpray",
               #   "EffectOrientedMCTSSpray"
                 ]
comparing_variables = [["teamsize_2","teamsize_3","teamsize_4","teamsize_5","teamsize_6"],
                       ["bound1_50","bound1_100","bound1_150","bound1_200","bound1_250","bound1_300","bound1_400","bound1_500","bound1_700","bound1_1000","bound1_1500"],
                    ["bound1_20","bound1_40","bound1_60","bound1_80","bound1_100","bound1_150","bound1_200","bound1_250","bound1_300"],
                    ["bound1_20","bound1_40","bound1_60","bound1_80","bound1_100","bound1_150","bound1_200","bound1_250","bound1_300","bound1_400","bound1_500","bound1_600","bound1_700"]]
seeds = [7,11,18,20,25,42,50,72,80,85]
# bound1_mcts = [1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000, 10000]
# bound1_SA = [100, 200, 300, 400, 500, 600]
team_sizes = [2,3,4,5,6]
# sourcenums = [2,4,6,8]
# change_intervals = [6,9,12]
```

```python
strategy_concentration = np.empty([len(strategy_name),len(comparing_variables[0]),len(comparing_variables[3])], dtype = float)
strategy_runtime = np.empty([len(strategy_name),len(comparing_variables[0]),len(comparing_variables[3])], dtype = float)
for m in range(len(strategy_name)):
    for i in range(len(comparing_variables[0])):
        for j in range(len(comparing_variables[3])):
            comparing_variable = comparing_variables[3][j] + comparing_variables[0][i]
            seed = 0
            team_size = team_sizes[i]
            sourcenum = 4
            sche_step = 18
            adaptive_step = 18
            strategy = strategy_name[m]

            pklname = f'SEED_{seed}_X{Setting.grid_x}_Y{Setting.grid_y}_VS{team_size}_TS{Setting.max_num_samples}_SS{sche_step}_AS{adaptive_step}_SN{sourcenum}_RS{Setting.replenish_speed}_WV{Setting.water_volume}'
            pkldir = '../outputs/' + strategy + '/' + comparing_variable + '/' + pklname + ".pkl"
            data = pypolo2.experiments.utilities.readpkl(pkldir)

            strategy_concentration[m,i,j] = np.mean(data['truth_env'][15:18])
            strategy_runtime[m,i,j] = data['runtime']
            # print(np.mean(data['truth_env'][15:18]))
            # print(data['runtime'])
```

```python
# 画柱状图，收敛时耗时对比。
# nondouple teamsize 2 3 4 5 6
# 268 279 1041 611 531
# smartspr teamsize 2 3 4 5 6
# 92 128 292 216 273 

# 固定污染源数量与车辆数相同时(稳定搜到):
# nondouple teamsize 2 3 4 5 6
# 441 591 2389 6823 3617
# smartspr teamsize 2 3 4 5 6
# 101 129 286 335 351

# 固定污染源数量与车辆数相同时(第一次搜到):
# nondouple teamsize 2 3 4 5 6
# 99 141 190 884 323
# smartspr teamsize 2 3 4 5 6
# 97 129 171 335 245

# 固定污染源数量与车辆数相同时(第二次搜到):
# nondouple teamsize 2 3 4 5 6
# 112 364 1008 1160 1263
# smartspr teamsize 2 3 4 5 6
# 101 134 286 411 268

import matplotlib.pyplot as plt
import numpy as np

# Data
teamsize = ['2', '3', '4', '5']
nondecoupled = [112, 364, 1008, 1160]
smarspr = [101, 134, 286, 411]

x = np.arange(len(teamsize))  
width = 0.35  

# plt.figure(figsize =(6,4),dpi=150)

fig, ax = plt.subplots(figsize =(6,4),dpi=150)
pm25_hatch = '//'
pm10_hatch = '\\\\'
edge_color = 'white'

for i in range(len(teamsize)):
    ax.bar(x[i] - width/2, nondecoupled[i], width, label='TRACT' if i == 0 else "",
           color='skyblue', hatch=pm25_hatch, edgecolor=edge_color)
    ax.bar(x[i] + width/2, smarspr[i], width, label='SmartSpr' if i == 0 else "",
           color='orange', hatch=pm10_hatch, edgecolor=edge_color)

ax.set_ylabel('Runtime (s)')
ax.set_xlabel('Sprinkler Number')
# ax.set_title('Overall Performance')
ax.set_xticks(x)
ax.set_ylim([0, 1500])
ax.set_xticklabels(teamsize)
ax.legend()

def autolabel(values, offset):
    for i, value in enumerate(values):
        fontweight = 'bold' if teamsize[i] == 'Ours' else 'normal'
        ax.annotate('{}'.format(value),
                    xy=(x[i] + offset, value),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight=fontweight)

autolabel(nondecoupled, -width/2)
autolabel(smarspr, width/2)

plt.legend(loc='upper center', bbox_to_anchor=(0.5, 1.12), ncol=2, frameon=False,columnspacing=0.6)

plt.rcParams['legend.fontsize'] = 11  # 图例字体大小
plt.rcParams['axes.labelsize'] = 12  # 坐标轴标签字体大小
plt.rcParams['xtick.labelsize'] = 11  # x轴刻度字体大小
plt.rcParams['ytick.labelsize'] = 11  # y轴刻度字体大小

fig.tight_layout()
filename = './outputs/png/performance_varnumsource.png'
plt.savefig(filename,bbox_inches = 'tight')

```
